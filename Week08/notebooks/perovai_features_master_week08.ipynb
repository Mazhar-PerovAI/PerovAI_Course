```python
import pandas as pd

type4 = pd.read_csv("../results/perovai_type4_thickness_week08.csv")
type5 = pd.read_csv("../results/perovai_type5_statistics_week08.csv")

print(len(type4), len(type5))
```

```python
features = []

stats = (
    type5.groupby(["HTL","Electrode"])["Vbr"]
    .agg(mean="mean", std="std", median="median")
    .reset_index()
)

for _, r in stats.iterrows():
    features.append({
        "HTL": r["HTL"],
        "Electrode": r["Electrode"],
        "feature": "Vbr_mean",
        "value": r["mean"]
    })
    features.append({
        "HTL": r["HTL"],
        "Electrode": r["Electrode"],
        "feature": "Vbr_std",
        "value": r["std"]
    })
    features.append({
        "HTL": r["HTL"],
        "Electrode": r["Electrode"],
        "feature": "Vbr_median",
        "value": r["median"]
    })
```

```python
for (htl, elec), g in type4.groupby(["HTL","Electrode"]):
    if g["thickness_nm"].nunique() > 1:
        slope = g.sort_values("thickness_nm")["Vbr"].diff().mean()
        features.append({
            "HTL": htl,
            "Electrode": elec,
            "feature": "Vbr_thickness_slope",
            "value": slope
        })
```

```python
features_df = pd.DataFrame(features)
features_df["architecture"] = "PIN"
features_df["bandgap_eV"] = 1.67

features_df.to_csv("../results/perovai_features_master_week08.csv", index=False)
print("Saved: ../results/perovai_features_master_week08.csv")
features_df.head()
```
