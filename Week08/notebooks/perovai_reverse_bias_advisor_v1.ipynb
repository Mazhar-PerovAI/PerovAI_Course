```python
import json
import pandas as pd   # ← this line is missing

features = pd.read_csv("../results/perovai_features_master_week08.csv")

with open("../results/knowledge/perovai_rules.json") as f:
    rules = json.load(f)

print("Loaded features:", features.shape)
print("Loaded rules:", len(rules))
```

```python
def reverse_bias_advisor(design):
    result = {"risks": [], "benefits": [], "recommendations": []}

    # filter features
    f = features[
        (features["HTL"] == design["HTL"]) &
        (features["Electrode"] == design["Electrode"])
    ]

    for _, row in f.iterrows():
        if row["feature"] == "Vbr_mean" and row["value"] > -5:
            result["risks"].append("Low reverse breakdown voltage")
            result["recommendations"].append("Use thicker HTL or more stable electrode")

        if row["feature"] == "Vbr_thickness_slope" and row["value"] < 0:
            result["benefits"].append("Thickness increases stability")

    return result
```

```python
test_design = {
    "architecture": "PIN",
    "bandgap_eV": 1.67,
    "Electrode": "Au",
    "HTL": "PTAA",
    "HTL_thickness_nm": 35,
    "ITO_planarization": True,
    "electron_blocking": True
}

reverse_bias_advisor(test_design)
```
